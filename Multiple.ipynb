```python
# 3. Create an Engine class with engine_type and a method start_engine(). 
# Create a Wheels class with number_of_wheels. Create Car that inherits from both and displays all details.

class Engine:
    def __init__(self, engine_type):
        self.engine_type = engine_type

    def start_engine(self):
        print("Start the engine")

class Wheels:
    def __init__(self, number_of_wheels):
        self.number_of_wheels = number_of_wheels

class Car(Engine, Wheels):
    def __init__(self, engine_type, number_of_wheels):
        Engine.__init__(self, engine_type)
        Wheels.__init__(self, number_of_wheels)

    def display(self):
        Engine.start_engine(self)
        print(f"Engine Type: {self.engine_type}")
        print(f"Number of Wheels: {self.number_of_wheels}")

obj = Car("Petrol", 4)
obj.display()
```

```python
# 6. Create Employee with name and salary. Create Bonus with bonus_amount. Create Payroll that calculates the employee's total salary including the bonus.

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Bonus:
    def __init__(self, bonus_amount):
        self.bonus_amount = bonus_amount

class Payroll(Employee, Bonus):
    def __init__(self, name, salary, bonus_amount):
        super().__init__(name, salary)
        Bonus.__init__(self, bonus_amount)

    def total_salary(self):
        return self.salary + self.bonus_amount

obj = Payroll("Shreya", 50000, 2500)
print(f"Name: {obj.name}")
print(f"Salary: {obj.salary}")
print(f"Bonus Amount: {obj.bonus_amount}")
print(f"Total Salary: {obj.total_salary()}")
```

```python
# 11. Create Theory to store theory marks and Practical to store practical marks. Create CourseResult that inherits from both and calculates total marks and percentage.

class Theory:
    def __init__(self, theory_1, theory_2, theory_3):
        self.theory_1 = theory_1
        self.theory_2 = theory_2
        self.theory_3 = theory_3

class Practical:
    def __init__(self, practical_1, practical_2, practical_3):
        self.practical_1 = practical_1
        self.practical_2 = practical_2
        self.practical_3 = practical_3

class CourseResult(Theory, Practical):
    def __init__(self, theory_1, theory_2, theory_3, practical_1, practical_2, practical_3):
        super().__init__(theory_1, theory_2, theory_3)
        Practical.__init__(self, practical_1, practical_2, practical_3)

    def total(self):
        return self.theory_1 + self.theory_2 + self.theory_3 + self.practical_1 + self.practical_2 + self.practical_3

    def percent(self):
        return (self.total() / 600) * 100

obj = CourseResult(85, 84, 95, 86, 78, 90)
print(f"Theory Marks of Subject 1: {obj.theory_1}")
print(f"Theory Marks of Subject 2: {obj.theory_2}")
print(f"Theory Marks of Subject 3: {obj.theory_3}")
print(f"Practical Marks of Subject 1: {obj.practical_1}")
print(f"Practical Marks of Subject 2: {obj.practical_2}")
print(f"Practical Marks of Subject 3: {obj.practical_3}")
print(f"Total Marks: {obj.total()}")
print(f"Percentage: {obj.percent():.2f} %")
```

```python
# 13. Create ShoppingCart with product prices and a method to calculate the total. Create Discount with a discount percentage. 
# Create FinalBill that calculates the final payable amount using both classes.

class ShoppingCart:
    def __init__(self, product_1, product_2, product_3):
        self.product_1 = product_1
        self.product_2 = product_2
        self.product_3 = product_3

    def total(self):
        return self.product_1 + self.product_2 + self.product_3

class Discount:
    def __init__(self, discount_percentage):
        self.discount_percentage = discount_percentage

class FinalBill(ShoppingCart, Discount):
    def __init__(self, product_1, product_2, product_3, discount_percentage):
        super().__init__(product_1, product_2, product_3)
        Discount.__init__(self, discount_percentage)

    def discount(self):
        return (self.total() * self.discount_percentage) / 100

obj = FinalBill(500, 200, 500, 10)
print(f"Price of Product 1: {obj.product_1}")
print(f"Price of Product 2: {obj.product_2}")
print(f"Price of Product 3: {obj.product_3}")
print(f"Total Amount: {obj.total()}")
print(f"Discount: {obj.discount_percentage}%")
print(f"Final Payable Amount: {obj.total() - obj.discount()}")
```

```python
# 14. Create Marks containing a student's marks. Create Attendance containing attendance percentage. 
# Create Eligibility that determines whether the student is eligible for an examination based on both marks and attendance.

class Marks:
    def __init__(self, marks):
        self.marks = marks

class Attendance:
    def __init__(self, attendance_percentage):
        self.attendance_percentage = attendance_percentage

class Eligibility(Marks, Attendance):
    def __init__(self, marks, attendance_percentage):
        super().__init__(marks)
        Attendance.__init__(self, attendance_percentage)

    def exam_eligibility(self):
        if self.marks >= 40  and self.attendance_percentage >= 40:
            print("Student is eligibile for examination")
        else:
            print("Student is not eligible for examination")

obj = Eligibility(80, 50)
print(f"Student Marks: {obj.marks}%")
print(f"Student Attendance Percentage: {obj.attendance_percentage}%")
obj.exam_eligibility()
```

```python
# 17. Create Academic with degree, CGPA, and specialization. Create Internship with company name, duration, and role. 
# Create Graduate that inherits from both and generates a complete profile.

class Academic:
    def __init__(self, degree, cgpa, specialization):
        self.degree = degree
        self.cgpa = cgpa
        self.specialization = specialization

class Internship:
    def __init__(self, company_name, duration, role):
        self.company_name = company_name
        self.duration = duration
        self.role = role

class Graduate(Academic, Internship):
    def __init__(self, degree, cgpa, specialization, company_name, duration, role):
        super().__init__(degree, cgpa, specialization)
        Internship.__init__(self, company_name, duration, role)

    def profile(self):
        print("----Academic Details----")
        print(f"Degree: {self.degree}")
        print(f"CGPA: {self.cgpa}")
        print(f"Specialization: {self.specialization}")
        print()
        print("----Internship Details----")
        print(f"Company Name: {self.company_name}")
        print(f"Duration: {self.duration}")
        print(f"Role: {self.role}")

obj = Graduate("B. E.", 4.5, "AIML", "VentCorp", 2, "IT Intern")
obj.profile()
```
