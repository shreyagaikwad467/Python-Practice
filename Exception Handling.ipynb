```python
# Write a program to handle ZeroDivisionError.

a = int(input("Enter 1st number: "))
b = int(input("Enter 2nd number: "))

try:
    a / b
except ZeroDivisionError:
    print("Division by zero is not possible")
else:
    print(f"{a} / {b} = {a / b}")
finally:
    print("Execuion completed")
```

```python
# 12. Write a program to read a file and handle FileNotFoundError and other possible exceptions.

try:
    file = open("Hello.txt", "r")
except FileNotFoundError:
    print("File not found")
except Exception as e:
    print("An error occurred:", e)
else:
    print(file.read())
    file.close()
finally:
    print("Execution Completed.")
```

```python
# 17. Create a custom exception InsufficientBalanceError for a bank withdrawal program.

class InsufficientBalanceError(Exception):
    pass

class Money:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance
        
    def display(self):
        print(f"Account Number: {self.account_number}")
        print(f"Balance: {self.balance}")
        
    def withdraw(self):
        withdraw_amount = float(input("Enter withdraw amount: "))

        if withdraw_amount > self.balance:
            raise InsufficientBalanceError("Insufficient Balance")
        
        self.balance = self.balance - withdraw_amount
        print(f"Balance after withdrawal: {self.balance}")
        
    def deposit(self):
        deposit_amount = float(input("Enter deposit amount: "))
        self.balance = self.balance + deposit_amount
        print(f"Balance after deposit: {self.balance}")

m = Money(1234, 50000)
m.display()
m.deposit()

try:
    m.withdraw()
except InsufficientBalanceError as e:
    print(e)
```

```python
# 18. Write a program using nested try-except blocks.

try:
    a = int(input("Enter 1st number: "))

    try:
        b = int(input("Enter 2nd number: "))
        div = a / b

    except ZeroDivisionError:
        print("Division by zero is not possible.")

    else:
        print(f"{a} / {b} = {div}")

except ValueError:
    print("Incorrect input!")
```

```python
# 19. Write a program demonstrating the complete flow of try → except → else → finally.

password = input("Enter password: ")

try:
    if password != "Shreya":
        raise Exception("Incorrect password!")
except Exception as e:
    print(e)
else:
    print("Access Granted!")
finally:
    print("Password verification completed.")
```

```python
# 20. Create a custom exception for invalid password length and handle it using try-except.

class PasswordLength(Exception):
    pass

password = input("Enter password: ")

try:
    if len(password) < 10:
        raise PasswordLength("Password must be at least 10 characters long.")
except PasswordLength as e:
    print(e)
else:
    print("Password length is okay.")
```
