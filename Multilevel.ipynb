```python
# 12. Shape contains color. Rectangle contains length and breadth and calculates area. Cuboid adds height and calculates volume.

class Shape:
    def __init__(self, colour):
        self.colour = colour

class Rectangle(Shape):
    def __init__(self, colour, length, breadth):
        super().__init__(colour)
        self.length = length 
        self.breadth = breadth 

    def area_Rectangle(self):
        return self.length * self.breadth

class Cuboid(Rectangle):
    def __init__(self, colour, length, breadth, height):
        super().__init__(colour, length, breadth)
        self.height = height 

    def volume_Cuboid(self):
        return self.length * self.breadth * self.height

shape = Cuboid("Red", 8, 5, 6)
print(f"Colour of the Shape: {shape.colour}")
print(f"Length of the Rectangle: {shape.length}")
print(f"Breadth of the Rectangle: {shape.breadth}")
print(f"Area of the Rectangle: {shape.area_Rectangle()}")
print(f"Height of the Cuboid: {shape.height}")
print(f"Volume of the Cuboid: {shape.volume_Cuboid()}")
```

```python
# 15. Account → deposit and withdraw. SavingsAccount → interest calculation. PremiumAccount → cashback calculation

class Account:
    def __init__(self, account_holder, balance, deposit_amount, withdraw_amount):
        self.account_holder = account_holder
        self.balance = balance 
        self.deposit_amount = deposit_amount
        self.withdraw_amount = withdraw_amount

    def deposit(self):
        self.balance += self.deposit_amount
        print(f"Deposit Successful. Current Balance: {self.balance}") 

    def withdraw(self):
        if self.withdraw_amount > self.balance:
            print("Insufficient Balance")

        elif self.balance - self.withdraw_amount < 1000:
            print("Withdrawal Failed. Minimum balance of 1000 should be maintained.")

        else:
            self.balance -= self.withdraw_amount
            print(f"Withdrawal Successful. Remaining Balance: {self.balance}")

class SavingsAccount(Account):
    def __init__(self, account_holder, balance, deposit_amount, withdraw_amount, rate_of_interest, time):
        super().__init__(account_holder, balance, deposit_amount, withdraw_amount)
        self.rate_of_interest = rate_of_interest
        self.time = time

    def interest(self):
        print(f"Interest: {(self.balance * self.rate_of_interest * self.time) / 100}")

class PremiumSavingsAccount(SavingsAccount):
    def __init__(self, account_holder, balance, deposit_amount, withdraw_amount, rate_of_interest, time, cashback):
        super().__init__(account_holder, balance, deposit_amount, withdraw_amount, rate_of_interest, time)
        self.cashback = cashback

    def calculate_cashback(self):
        cashback_amount = self.deposit_amount * self.cashback / 100
        print(f"Calculate: {cashback_amount}")

a = PremiumSavingsAccount("Shreya", 10000, 2000, 1500, 5, 2, 5)
print(f"Account Holder: {a.account_holder}")
print(f"Balance: {a.balance}")
print(f"Deposit Amount: {a.deposit_amount}")
a.deposit()
print(f"Withdraw Amount: {a.withdraw_amount}")
a.withdraw()
print(f"Rate of Interest: {a.rate_of_interest}%")
print(f"Time: {a.time} years")
a.interest()
print(f"Cashback: {a.cashback}%")
a.calculate_cashback()
```

```python
# 16. Vehicle: brand, model. Car: number of doors. ElectricCar: battery capacity. Add a method in ElectricCar to calculate range assuming 1 kWh = 5 km.

class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

class Car(Vehicle):
    def __init__(self, brand, model, no_of_doors):
        super().__init__(brand, model)
        self.no_of_doors = no_of_doors

class ElectricCar(Car):
    def __init__(self, brand, model, no_of_doors, battery_capacity):
        super().__init__(brand, model, no_of_doors)
        self.battery_capacity = battery_capacity

    def calculate_range(self):
        return self.battery_capacity * 5

obj = ElectricCar("Maruti Suzuki", "Baleno", 4, 20)
print(f"Brand: {obj.brand}")
print(f"Model: {obj.model}")
print(f"Number of Doors: {obj.no_of_doors}")
print(f"Battery Capacity: {obj.battery_capacity} kWh")
print(f"Range: {obj.calculate_range()} km")
```

```python
# 17. Student contains name and roll number. Exam contains marks for five subjects. Result calculates:
# Total
# Percentage
# Grade

class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

class Exam(Student):
    def __init__(self, name, roll_no, s1, s2, s3, s4, s5):
        super().__init__(name, roll_no)
        self.s1 = s1
        self.s2 = s2
        self.s3 = s3
        self.s4 = s4
        self.s5 = s5

class Result(Exam):
    def __init__(self, name, roll_no, s1, s2, s3, s4, s5):
        super().__init__(name, roll_no, s1, s2, s3, s4, s5)

    def total(self):
        return self.s1 + self.s2 + self.s3 + self.s4 + self.s5

    def percentage(self):
        return ((self.s1 + self.s2 + self.s3 + self.s4 + self.s5) / 500) * 100

    def grade(self):
        if ((self.s1 + self.s2 + self.s3 + self.s4 + self.s5) / 500) * 100 >= 80:
            print("Grade A")

        elif ((self.s1 + self.s2 + self.s3 + self.s4 + self.s5) / 500) * 100 >= 60:
            print("Grade: B")

        elif ((self.s1 + self.s2 + self.s3 + self.s4 + self.s5) / 500) * 100 >= 40:
            print("Grade: C")

        else:
            print("Grade: D")

r = Result("Shreya", 42, 80, 90, 75, 85, 90)
print(f"Name: {r.name}")
print(f"Roll No.: {r.roll_no}")
print(f"Marks of Subject 1: {r.s1}")
print(f"Marks of Subject 2: {r.s2}")
print(f"Marks of Subject 3: {r.s3}")
print(f"Marks of Subject 4: {r.s4}")
print(f"Marks of Subject 5: {r.s5}")
print(f"Total Marks: {r.total()}")
print(f"Percentage: {r.percentage():.2f}%")
r.grade()
```

```python
# 19. Person: name, age. Student: roll number, marks. CollegeStudent: college name, course. 
# Create a method that determines whether the student has passed based on a percentage of 40% or above.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, roll_no, sub1, sub2, sub3):
        super().__init__(name, age)
        self.roll_no = roll_no
        self.sub1 = sub1
        self.sub2 = sub2
        self.sub3 = sub3 

class CollegeStudent(Student):
    def __init__(self, name, age, roll_no, sub1, sub2, sub3, college_name, course):
        super().__init__(name, age, roll_no, sub1, sub2, sub3)
        self.college_name = college_name
        self.course = course

    def pass_fail(self):
        self.percent = ((self.sub1 + self.sub2 + self.sub3) / 300) * 100
        print(f"Percentage: {self.percent:.2f}%")
        if self.percent >= 40:
            print("Pass")
        else:
            print("Fail")

obj = CollegeStudent("Shreya", 20, 42, 85, 96, 80, "SPPU", "AIML")
print(f"Name: {obj.name}")
print(f"Age: {obj.age}")
print(f"Roll No.: {obj.roll_no}")
print(f"College Name: {obj.college_name}")
print(f"Course: {obj.course}")
print(f"Marks of Subject 1: {obj.sub1}")
print(f"Marks of Subject 2: {obj.sub2}")
print(f"Marks of Subject 3: {obj.sub3}")
obj.pass_fail()
```

```python
# 20. Create a three-level hierarchy:
# University → Department → Professor
# University: university name, location
# Department: department name, HOD
# Professor: professor name, subject, experience
# Use constructors and super() at every level. Finally, create a Professor object and display all details.

class University:
    def __init__(self, university_name, location):
        self.university_name = university_name
        self.location = location

class Department(University):
    def __init__(self, university_name, location, department_name, hod):
        super().__init__(university_name, location)
        self.department_name = department_name
        self.hod = hod

class Professor(Department):
    def __init__(self, university_name, location, department_name, hod, professor_name, subject, experience):
        super().__init__(university_name, location, department_name, hod)
        self.professor_name = professor_name
        self.subject = subject
        self.experience = experience

prof = Professor("SPPU", "Pune", "AIML", "Mr. Kumar", "Mrs. Yadav", "Deep Learning", "5 Years")
print(f"University Name: {prof.university_name}")
print(f"University Location: {prof.location}")
print(f"Department Name: {prof.department_name}")
print(f"Department HOD: {prof.hod}")
print(f"Professor Name: {prof.professor_name}")
print(f"Subject: {prof.subject}")
print(f"Experience: {prof.experience}")
```
