```python
# 5. Create a Shape class with an attribute color. Create a Rectangle class that inherits from Shape and has length and breadth. Calculate the area.

class Shape:
    def __init__(self, colour):
        self.colour = colour

class Rectangle(Shape):
    def __init__(self, colour, length, breadth):
        super().__init__(colour)
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

    def display(self):
        print(f"Colour: {self.colour}")
        print(f"Length of Rectangle: {self.length}")
        print(f"Breadth of Rectangle: {self.breadth}")
        print(f"Area of Rectangle: {self.area()}")

r = Rectangle("Red", 10, 20)
r.display()
```

```python
# 8. Create a Student class with name and roll_no. 
# Create an ExamStudent class that adds marks for three subjects. 
# Calculate and display the total and percentage.

class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

    def display_stud(self):
        print(f"Name: {self.name}")
        print(f"Roll No.: {self.roll_no}")

class ExamStudent(Student):
    def __init__(self, name, roll_no, sub1, sub2, sub3):
        super().__init__(name, roll_no)
        self.sub1 = sub1
        self.sub2 = sub2
        self.sub3 = sub3

    def display_marks(self):
        print(f"Marks of English: {self.sub1}")
        print(f"Marks of Maths: {self.sub2}")
        print(f"Marks of Science: {self.sub3}")

    def total_percentage(self):
        total = self.sub1 + self.sub2 + self.sub3
        print(f"Total Marks: {total}")
        print(f"Percentage: {(total / 300) * 100:.2f}%")

stud1 = ExamStudent("Alex", 50, 98, 80, 75)
stud1.display_stud()
stud1.display_marks()
stud1.total_percentage()
```

```python
# 12. Create an Account class with account_holder and balance. Create a CurrentAccount class with a withdraw() method. Prevent withdrawal if the amount exceeds the balance.

class Account:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def display(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Current Balance: {self.balance}")

class CurrentAccount(Account):

    def withdraw(self, w):
        print(f"Withdrawal Amount: {w}")
        if w > self.balance:
            print("Insufficient Balance")
        elif self.balance - w < 1000:
            print("Minimum balance of 1000 is required. Cannot withdraw.")
        else:
            self.balance -= w
            print(f"Balance after Withdrawal: {self.balance}")

ca = CurrentAccount("Shreya", 2000)
ca.display()
ca.withdraw(200000)
ca.withdraw(2000)
ca.withdraw(200)
```

```python
# 16. Create a Vehicle class with brand and model. Create an ElectricCar class that adds battery_capacity. 
# Add a method that calculates the estimated range assuming 1 km per kWh.
# Estimated Range = Battery Capacity × Range per kWh

class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

class ElectricCar(Vehicle):
    range_per_kWh = 1

    def __init__(self, brand, model, battery_capacity):
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity

    def display(self):
        print(f"Brand: {self.brand}")
        print(f"Model: {self.model}")
        print(f"Battery Capacity: {self.battery_capacity} kWh")

    def estimated_range(self):
        return self.battery_capacity * self.range_per_kWh

ec = ElectricCar("Maruti Suzuki", "Swift", 25)
ec.display()
print(f"Estimated Range: {ec.estimated_range()} km")
```

```python
# 18. Create a Student class containing name and roll_no. Create a Result class that inherits from Student and accepts marks for five subjects. Add methods to calculate:
# Total marks
# Percentage
# Pass/Fail status

class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

class Result(Student):
    def __init__(self, name, roll_no, s1, s2, s3, s4, s5):
        super().__init__(name, roll_no)
        self.s1 = s1
        self.s2 = s2
        self.s3 = s3
        self.s4 = s4
        self.s5 = s5

    def display(self):
        print(f"Name: {self.name}")
        print(f"Roll No.: {self.roll_no}")
        print(f"English Marks: {self.s1}")
        print(f"Science Marks: {self.s2}")
        print(f"Maths Marks: {self.s3}")
        print(f"SST Marks: {self.s4}")
        print(f"Marathi Marks: {self.s5}")
    
    def total_marks(self):
        return self.s1 + self.s2 + self.s3 + self.s4 + self.s5

    def percentage(self):
        return ((self.s1 + self.s2 + self.s3 + self.s4 + self.s5) / 500) * 100

    def pass_fail(self):
        if (self.s1 + self.s2 + self.s3 + self.s4 + self.s5) / 500 * 100 >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")

r = Result("Shreya", 42, 50, 25, 45, 36, 42)
r.display()
print(f"Total Marks: {r.total_marks()}")
print(f"Percentage: {r.percentage()}%")
r.pass_fail()

print()

r1 = Result("Meena", 43, 80, 72, 82, 65, 58)
r1.display()
print(f"Total Marks: {r1.total_marks()}")
print(f"Percentage: {r1.percentage():.2f}%")
r1.pass_fail()
```

```python
# 19. Create a BankAccount class with account_holder and balance, along with deposit() and withdraw() methods. 
# Create a PremiumAccount class that inherits from it and adds a cashback() method that returns 5% of the deposited amount.

class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def display(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Current Balance: {self.balance}")


    def deposit(self):
        self.d = float(input("Enter Deposit Amount: "))
        self.balance += self.d
        print(f"Balance after Depositing: {self.balance}")

    def withdraw(self):
        w = float(input("Enter Withdrawal Amount: "))
        if w > self.balance:
            print(f"Insufficient Balance")
        elif self.balance - w < 1000:
            print("Minimum balance of 1000 should be maintained. Cannot Withdraw.")
        else:
            self.balance -= w
            print(f"Balance after Withdrawal: {self.balance}")

class PremiumAccount(BankAccount):
    def cashback(self):
        return self.d * 0.05

pa1 = PremiumAccount("Shreya", 2000)
pa1.display()
pa1.deposit()
print(f"Cashback: {pa1.cashback()}")
pa1.withdraw()
```
