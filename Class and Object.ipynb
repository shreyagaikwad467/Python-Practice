# Classes and Objects

## Class
    - A class is a blueprint for creating objects. 
    - It defines the attributes (data) and methods (functions) that the objects will have.
    - Classes are created using the class keyword.
    - Attributes are public by default in Python and can be accessed using the dot (.) operator.

## Object
    - An object is an instance of a class. 
    - It represents a real-world entity and has its own attributes (data) and methods (functions) defined by the class.
    - Objects are created from a class.
    - Each object has its own state (attribute values) and can access methods using the dot (.) operator.

```python
# 1. Create a Student class with name and age attributes. Create an object and display them.

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        # Store the value in the object.

    def printStudent(self):
        print(f"Name of the student: {self.name}")
        print(f"Age: {self.age}")
        # Access the value stored in the object.

stud1 = Student("Shreya", 20)
stud1.printStudent()
```

```python
# 2. Create a Car class with brand and model attributes. Create an object and display the details.

class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def displayCar(self):
        print(f"Car Brand: {self.brand}")
        print(f"Car Model: {self.model}")

car = Car("Maruti Suzuki", "Baleno")
car.displayCar()
```

```python
# 3. Create a Person class with name and age. Add a method display() to print the details.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

p = Person("Omkar", 21)
p.display()
```

```python
# 4. Create a Rectangle class with length and breadth. Add a method to calculate the area.

class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area_of_Rectangle(self):
        self.area = self.length * self.breadth

    def display(self):
        print(f"Length of the rectangle: {self.length}")
        print(f"Breadth of the rectangle: {self.breadth}")
        print(f"Area of the rectangle: {self.area}")

a = Rectangle(10, 12)
a.area_of_Rectangle()
a.display()
```

```python
# 5. Create a Circle class with radius. Add a method to calculate the area.

pi = 3.14
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area_of_Circle(self):
        self.area = pi * (self.radius**2)

    def display(self):
        print(f"Radius of the circle: {self.radius}")
        print(f"Area of the circle: {self.area}")

c = Circle(4)
c.area_of_Circle()
c.display()
```

```python
# 6. Create a Calculator class with methods for addition, subtraction, multiplication, and division.

class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def addition(self):
        print(f"{self.num1} + {self.num2} = {self.num1 + self.num2}")

    def subtraction(self):
        print(f"{self.num1} - {self.num2} = {self.num1 - self.num2}")

    def multiplication(self):
        print(f"{self.num1} * {self.num2} = {self.num1 * self.num2}")

    def division(self):
        try:
            print(f"{self.num1} / {self.num2} = {self.num1 / self.num2}")
        except ZeroDivisionError:
            print("Division by zero is not possible")

cal = Calculator(8, 0)
cal.addition()
cal.subtraction()
cal.multiplication()
cal.division()

# Tip : For calculator programs, include ZeroDivisionError
```

```python
# 7. Create a BankAccount class with account_number and balance. Add methods to deposit and withdraw money.

class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance

    def previous(self):
        print(f"Account Number: {self.account_number}")
        print(f"Balance: {self.balance}")

    def deposit(self):
        deposit_amount = float(input("Enter the amount to be deposited: "))
        self.balance = self.balance + deposit_amount
        print(f"Balance after depositing money: {self.balance}")

    def withdraw(self):
        withdraw_amount = float(input("Enter the amount to be withdrawn: "))
        if(withdraw_amount > self.balance):
            print("Insufficient Balance!")
        else:
            self.balance = self.balance - withdraw_amount
            print(f"Balance after withdrawing money: {self.balance}")

ba = BankAccount(123456, 20000)
ba.deposit()
```

```python
# 8. Create a Book class with title and author. Create two objects and display their details.

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def display(self):
        print(f"Title of the book: {self.title}")
        print(f"Author of the book: {self.author}")

b1 = Book("The Seven Husbands of Evelyn Hugo", "Taylor Jenkins Reid")
b1.display()
b2 = Book("The End of Us", "Collen Hoover")
b2.display()
```

```python
# 10. Create a Student class with name and marks. Add a method to check whether the student passed or failed.

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Name of the Student: {self.name}")
        print(f"Marks: {self.marks}")
        
    def result(self):
        if(self.marks >= 40):
            print("Pass")
        else:
            print("Fail")

stud = Student("Omkar", 90)
stud.display()
stud.result()
```
